```python
import pandas as pd
import requests
from math import ceil
```

```python
BASE_URL="http://20.228.134.116:8000/volcan"
MAX_LIMIT_ORDENES=200

#SOLICITUD DE METADATA 
res = requests.get(F"{BASE_URL}/lotes", params={'limit':1}).json()
# EXTRAER TOTAL DE ITEMS
total_ordenes: int = res ["total"]

#CALCULAR CANTIDAD DE batchs
num_batchs = ceil(total_ordenes / MAX_LIMIT_ORDENES)

print(num_batchs) 
```

```python
acopio_dfs = []

for i in range(num_batchs):
    offset = i * MAX_LIMIT_ORDENES
    params = {
        "offset": offset,
        "limit": MAX_LIMIT_ORDENES
    }
    print(f"{i}| obteniendo datos con: {params}")

    res = requests.get(F"{BASE_URL}/lotes", params=params).json()
    print(res['total'])

    #crear dataframe
    batch_df = pd.DataFrame(res["items"])

    acopio_dfs.append(batch_df)

acopio_df = pd.concat(acopio_dfs)   
print(acopio_df.info())
```
